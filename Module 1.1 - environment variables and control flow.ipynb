## 2. Variables, values, and types

A **variable** is a name that refers to a value. The assignment operator, `=`, saves the value on the right using the name on the left. It does not mean mathematical equality.

Use descriptive `snake_case` names. Avoid names such as `list`, `str`, `input`, or `sum`, because those names already refer to useful Python tools.

```python
customer_name = '<Ali>'         # str: text
orders_this_month = 12                       # int: whole number
average_order_value = 48.75                  # float: decimal number
is_active_customer = True                    # bool: True or False

print(f"Customer Name: {customer_name} ||", 
      f"Orders This Month: {orders_this_month} ||", 
      f"Average Order Value: {average_order_value} ||", 
      f"Is Active Customer: {is_active_customer}")
print(f"Type of customer_name: {type(customer_name)}")
print(f"Type of orders_this_month: {type(orders_this_month)}")
print(f"Type of average_order_value: {type(average_order_value)}")
print(f"Type of is_active_customer: {type(is_active_customer)}")
```

### Assignment versus comparison

- `total = 10` assigns `10` to `total`.
- `total == 10` asks whether `total` is equal to `10`; it evaluates to either `True` or `False`.

```python
total = 10
print(total == 25)  # This is a boolean expression that evaluates to True or False. In this case, it will print True because total is equal to 10.
print(total != 5)  # This is a boolean expression that evaluates to True or False. In this case, it will print False because total is equal to 10.

# Python evaluates the right side first, then updates the variable.
total = total + 5   # This is an assignment statement that updates the value of total by adding 5 to its current value. Running it multiple times will keep increasing the value of total by 5 each time.
print(total)
```

## 3. Arithmetic and numeric types

| Operator | Meaning | Example result |
|---|---|---|
| `+` | Addition | `5 + 2` → `7` |
| `-` | Subtraction | `5 - 2` → `3` |
| `*` | Multiplication | `5 * 2` → `10` |
| `/` | Division; returns a float | `5 / 2` → `2.5` |
| `//` | Floor division | `5 // 2` → `2` |
| `%` | Remainder (modulo) | `5 % 2` → `1` |
| `**` | Exponentiation | `5 ** 2` → `25` |

Use `round(number, digits)` when presenting a numerical result. Keep full precision during calculations unless presentation requires rounding.

```python
revenue = 12_500.50                                  # notice the use of underscore to make the number more readable
orders = 286                                         # number of orders
average_revenue_per_order = revenue / orders         # calculate average revenue per order

print('Average revenue per order:', round(average_revenue_per_order, 2))  # round() in this case rounds the average revenue per order to 2 decimal places.
print('Full boxes:', orders // 12)                                        # the // operator performs floor division, so it returns the number of full boxes (12 orders per box) without any remainder.
print('Items left after full boxes:', orders % 12)                        # the % operator performs modulo division, so it returns the number of items left after filling full boxes.
print('Is the order count even?', orders % 2 == 0)                        # the % operator in this case can be used to check if a number is even or odd.
```

## 4. Strings and cleaning text

A **string** (`str`) is text wrapped in quotes. Real-world data often contains extra spaces, inconsistent capitalization, or several pieces of information in one field.

Useful methods:
- `.strip()` removes whitespace at the beginning and end.
- `.title()` converts words to title case.
- `.lower()` converts text to lowercase.
- `.split(separator)` divides a string into pieces.
- f-strings place values inside text: `f'Name: {name}'`.

String methods return a new value. Save that result when you need to reuse it.

```python
raw_name = '   cHarLeY hAnAnIa   '               # str: text with extra spaces and inconsistent capitalization
clean_name = raw_name.strip().title()            # strip() removes leading and trailing whitespace, and the title() function capitalizes the first letter of each word in the string.

print(clean_name)                                # Print the cleaned-up version of the name.
print(f'Welcome, {clean_name}!')                 # Print a welcome message that includes the cleaned-up name.

raw_tags = 'python, pandas, visualization'       # str: text with tags separated by commas
tags = raw_tags.split(', ')                      # split() splits the string into a list of substrings based on the specified delimiter (in this case, ', ').
print(tags)                                      # Print the list of tags.
print(f'First tag: {tags[0]}')                   # Print the first tag in the list.
```

## 5. Functions: reusable blocks of logic

A function gives a name to a repeatable task. Parameters are inputs to the function. `return` sends a result back to the place where the function was called.

A function should normally **return data** when later code needs that result. It can **print a message** when its main purpose is communicating with a person.

```python
def calculate_average(total_value, number_of_items):       # We're defining a function named calculate_average that takes two parameters: total_value and number_of_items.
    """Return the average value per item."""               # This is a comment (docstring) that describes what the function does. It returns the average value per item.
    return total_value / number_of_items                   # Calculates the average value by dividing total_value by number_of_items, returning the result.

average = calculate_average(12_500.50, 286)                # Call the calculate_average function (ABOVE) with total_value of 12,500.50 and number_of_items of 286, storing the result in the variable average.
print(round(average, 2))                                   # Print the average value rounded to 2 decimal places.
```

```python
def format_currency(amount, currency='EUR'):                                # We're defining a function named format_currency that takes two parameters: amount and currency. The currency parameter has a default value of 'EUR'.
    """Format an amount with two decimal places and a currency code."""     
    return f'{amount:,.2f} {currency}'                                      # Returns a formatted string that includes the amount with two decimal places and the specified currency code. The :,.2f format specifier formats the number with commas as thousands separators and two decimal places. More formats can be found here: https://docs.python.org/3/library/string.html#format-specification-mini-language

print(format_currency(1234.5))                                              # Call the format_currency function with an amount of 1234.5 and the default currency of 'EUR'.
print(format_currency(1234.5, 'CHF'))                                       # Call the format_currency function with an amount of 1234.5 and the currency of 'CHF'.
```

### Scope: local and global names
A variable created inside a function is normally **local** to that function. Prefer passing values through parameters and returning results instead of relying on global variables. That makes code easier to test and reuse.

```python
tax_rate = 0.20                                           # A global name; avoid changing it inside functions.

def add_tax(price, rate):                                 # The function takes two parameters: price and rate.
    final_price = price * (1 + rate)                      # Calculate the final price by multiplying the original price by (1 + rate). This effectively adds the tax to the original price.
    return final_price                                    # Return the final price after tax.

print(format_currency(add_tax(100, tax_rate)))            # Format and print the final price of an item that costs 100 with the tax rate applied.
```

## 6. Conditionals and Boolean logic

Conditionals let a program choose an action based on a condition. Every condition evaluates to `True` or `False`.

| Operator | Meaning |
|---|---|
| `==` | Equal to |
| `!=` | Not equal to |
| `>` / `<` | Greater than / less than |
| `>=` / `<=` | Greater than or equal to / less than or equal to |
| `and` | Both conditions must be true |
| `or` | At least one condition must be true |
| `not` | Reverses a Boolean value |

```python
# We use control flow to make decisions in our code. The if statement allows us to execute a block of code only if a certain condition is true. We can also use elif (short for "else if") and else to handle multiple conditions.

score = 82
attendance_rate = 0.91

if score >= 85 and attendance_rate >= 0.80:    # note the ":" at the end of the line, which indicates that the following indented block of code is part of the if statement. The "and" operator is used to combine two conditions, so both must be true for the block of code to execute.
    result = 'Distinction'                     # This line is executed if the condition in the if statement is true. It assigns the string 'Distinction' to the variable result.
elif score >= 70:                              # This line checks another condition if the previous if statement was false. If score is greater than or equal to 70, this block of code will execute.
    result = 'Pass'                            # This line is executed if the condition in the elif statement is true. It assigns the string 'Pass' to the variable result.
else:                                          # This line is executed if none of the previous conditions were true. It serves as a catch-all for any other cases.
    result = 'Needs further practice'          # This line is executed if the else block is reached. It assigns the string 'Needs further practice' to the variable result.

print(result)
```

### Structural pattern matching
`match` / `case` is useful when one value can map to several named options. It requires Python 3.10 or later. Use `if` / `elif` when you are evaluating ranges or more complex conditions.

```python
# match statement is a control flow structure that allows you to compare a value against multiple patterns and execute code based on which pattern matches.
# It is similar to a switch statement in other programming languages.
learning_route = 'data_analysis'

match learning_route:                                                          # match compared to if/elif/else: match is more concise and easier to read when dealing with multiple conditions. It allows you to define different cases for different values of a variable, making the code cleaner and more organized.
    case 'python_foundations':                                                 # This line defines a case for when learning_route is equal to 'python_foundations'.
        message = 'Start with variables, functions, and control flow.'         # This line is executed if the case matches. It assigns the string 'Start with variables, functions, and control flow.' to the variable message.
    case 'data_analysis':                                                      # This line defines a case for when learning_route is equal to 'data_analysis'.
        message = 'Next, we will use pandas for tabular data.'                 # This line is executed if the case matches. It assigns the string 'Next, we will use pandas for tabular data.' to the variable message.
    case 'machine_learning':                                                   # This line defines a case for when learning_route is equal to 'machine_learning'.
        message = 'We will prepare data and evaluate predictive models.'       # This line is executed if the case matches. It assigns the string 'We will prepare data and evaluate predictive models.' to the variable message.
    case _:                                                                    # This line defines a default case that will match any value of learning_route that doesn't match the previous cases.
        message = 'Choose a listed learning route.'                            # This line is executed if the default case is reached. It assigns the string 'Choose a listed learning route.' to the variable message.

print(message)
```

## Code-Along 1.1.2 — Conditional grading and route tool

### Goal
Use a function, range-based grading logic, and `match` / `case` to produce an actionable message.

```python
def assign_grade(score):
    """Return a letter grade for a score from 0 through 100."""
    if score >= 90:
        return 'A'
    elif score >= 80:
        return 'B'
    elif score >= 70:
        return 'C'
    elif score >= 60:
        return 'D'
    return 'F'

learner_score = 83
grade = assign_grade(learner_score)

match grade:
    case 'A' | 'B':
        next_step = 'Move on to the extension challenge.'
    case 'C':
        next_step = 'Complete the core practice before continuing.'
    case 'D' | 'F':
        next_step = 'Attend office hours and repeat the guided practice.'
    case _:
        next_step = 'Check the score input.'

print(f'Score: {learner_score} | Grade: {grade}')
print(next_step)
```

### Challenge
Modify `assign_grade()` so that scores below 0 or above 100 return `'Invalid score'`. Then update the `match` statement to provide a suitable message.

```python
def assign_grade(score):
    if score < 0 or score > 100:
        return 'Invalid score'
    if score >= 90:
        return 'A'
    elif score >= 80:
        return 'B'
    elif score >= 70:
        return 'C'
    elif score >= 60:
        return 'D'
    return 'F'

for score in [-1, 0, 59, 60, 70, 80, 90, 101]:
    grade = assign_grade(score)

    match grade:
        case 'A' | 'B':
            message = 'Move on to the extension challenge.'
        case 'C':
            message = 'Complete the core practice.'
        case 'D' | 'F':
            message = 'Repeat the guided practice.'
        case 'Invalid score':
            message = 'Enter a score from 0 to 100.'

    print(f'{score}: {grade} — {message}')
```

## 7. Iteration: `for` and `while` loops

Use a `for` loop when you know what collection or sequence you want to visit. Use a `while` loop when repetition should continue until a condition changes.

- `range(5)` produces numbers from 0 through 4.
- `_` is a conventional name when the loop counter itself is not needed.
- `break` immediately exits a loop.
- `continue` skips the rest of the current iteration and begins the next one.

```python
for day_number in range(1, 6):
    print(f'Day {day_number}: practise Python for 30 minutes.')

print('Practice week complete.')
```

```python
daily_sales = [125, 0, 210, -15, 180]
valid_total = 0

for sale in daily_sales:
    if sale < 0:
        print(f'Skipping invalid value: {sale}')
        continue
    valid_total += sale

print(f'Valid sales total: {valid_total}')
```

### Avoiding infinite loops
A `while` loop needs a condition that will eventually become false, or it needs a deliberate `break`. If a notebook cell appears to run forever, use the stop button or **Kernel → Interrupt**.

```python
attempts_remaining = 3

while attempts_remaining > 0:
    print(f'You have {attempts_remaining} attempt(s) remaining.')
    attempts_remaining -= 1  # This update ensures the loop ends.

print('The loop has ended safely.')
```

## Code-Along 1.1.3 — Defensive prompt loop

### Goal
Build a loop that accepts only an allowed answer. We will start with a testable function, then show the interactive version.

This is a first step toward data-quality validation: before analysis, we check whether values meet the rules we expect.

```python
def normalise_yes_no(response):
    """Return 'yes' or 'no' for accepted text; otherwise return None."""
    cleaned_response = response.strip().lower()

    if cleaned_response in ('yes', 'y'):
        return 'yes'
    if cleaned_response in ('no', 'n'):
        return 'no'
    return None

test_responses = [' perhaps ', 'Y', 'NO']

for response in test_responses:
    validated_response = normalise_yes_no(response)
    print(f'{response!r} -> {validated_response!r}')
```

## Independent practice — mini data-quality report

You receive a text field containing a reported number of hours studied. Create a function called `study_status(hours_text)` that:
1. Removes extra whitespace.
2. Converts the text to a number.
3. Returns `'Invalid'` if the number is negative.
4. Returns `'On track'` when the number is 5 or more.
5. Returns `'Needs more practice'` otherwise.

**Stretch:** In Module 1.2, update your solution so that non-numeric text such as `'five'` is handled with `try` / `except`.

```python
def study_status(hours_text):
    hours = float(hours_text.strip())

    if hours < 0:
        return 'Invalid'
    elif hours >= 5:
        return 'On track'
    else:
        return 'Needs more practice'

for test_value in [' 6 ', '4.5', '0', '-2']:
    print(f'{test_value!r} -> {study_status(test_value)}')
```

Read the error message before changing code. A traceback usually identifies the line where Python could no longer continue. Common early errors include:

- `NameError`: a name has not been defined, often because of a spelling mismatch or cells run out of order.

- `ValueError`: a conversion could not be completed, such as `int('five')`.
- `IndentationError`: Python expected a consistently indented code block.
- `SyntaxError`: Python could not parse the code; check quotes, parentheses, colons, and spelling.
